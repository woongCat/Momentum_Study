```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random

# 사용자 에이전트 설정 (웹사이트에서 차단 방지)
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

def get_melon_chart_songs(chart_url, limit=30):
    """멜론 차트에서 노래 제목과 가수, 노래 ID를 가져옵니다."""
    response = requests.get(chart_url, headers=headers)
    soup = BeautifulSoup(response.text, 'html.parser')
    
    songs = []
    for tr in soup.select('table tbody tr')[:limit]:
        song_id = tr.get('data-song-no')
        title = tr.select_one('div.ellipsis.rank01 a').text.strip()
        artist = tr.select_one('div.ellipsis.rank02 a').text.strip()
        songs.append({
            'title': title,
            'artist': artist,
            'song_id': song_id
        })
    return songs

def get_lyrics(song_id):
    """노래 ID로 가사 크롤링"""
    url = f'https://www.melon.com/song/detail.htm?songId={song_id}'
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # 가사 추출
    lyrics_div = soup.select_one('div.lyric')
    if lyrics_div:
        lyrics = lyrics_div.get_text(separator='\n').strip()
        return lyrics
    return "가사를 찾을 수 없습니다."
```

```python

# 멜론 실시간 차트에서 노래 목록 가져오기
chart_url = 'https://www.melon.com/chart/index.htm'
songs = get_melon_chart_songs(chart_url, limit=100)  # 상위 5곡만 가져오기

# 각 노래의 가사 수집
for song in songs:
    song_id = song['song_id']
    print(f"\n가져오는 중: {song['title']} - {song['artist']}")
    
    # 가사 가져오기
    lyrics = get_lyrics(song_id)
    song['lyrics'] = lyrics
    
    # 웹사이트에 부하를 주지 않기 위해 잠시 대기
    time.sleep(random.uniform(1, 2))

# 결과를 데이터프레임으로 변환
df_songs = pd.DataFrame(songs)

# 결과 확인
print("\n수집된 노래 정보:")
display(df_songs[['title', 'artist']])

# 가사 확인
for idx, row in df_songs.iterrows():
    print(f"\n=== {row['title']} - {row['artist']} ===")
    print(row['lyrics'][:200] + "...")  # 일부만 출력
```

```python
df_songs.to_csv("songs.xlsx", index=False)
```
